```python
# SEE OUTLIERS THROUGH IQR
```

```python
numerical_cols = ['study_time_hours','attendance_percent','sleep_hours','previous_grade', 'final_exam_score']
```

```python
for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    print(f"\n{col}")

    print("Q1:", Q1)

    print("Q3:", Q3)

    print("IQR:", IQR)
    print("Lower bound:", lower_bound)
    print("Upper bound:", upper_bound)
```

```python
# CHECK ACTUAL OUTLIER VALUES
```

```python
for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ][col]

    print(f"\n{col}")
    print("Number of outliers:", len(outliers))
    print("Outlier values:", outliers.tolist())
```

```python
# HANDLED OUTLIERS
```

```python
for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df[col] = df[col].clip(
        lower=lower_bound,
        upper=upper_bound
    )
```

```python
for col in numerical_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ][col]

    print(f"\n{col}")
    print("Number of outliers:", len(outliers))
```

```python
# SEPARATE FEATURES AND TARGET
```

## OUR TARGET IS FINAL GRADE

```python
X = df.drop(columns=['student_id', 'final_grade'])
y = df['final_grade']
```

```python
# CHECK SHAPES
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
# SPLIT THE DATA INTO TRAINING AND TESTING
```

```python
X_train, X_test, y_train, y_test = train_test_split( X, y, test_size=0.20, random_state=42)
```

```python
# CHECK SIZES
```

```python
print("X_train:", X_train.shape)

print("X_test:",X_test.shape)

print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
# SEPARATE CATEGORICAL AND NUMERICAL
```

```python
numerical_features = ["study_time_hours","attendance_percent","sleep_hours","previous_grade","final_exam_score"]

categorical_features = ["gender","parental_education","internet_access","extracurricular_activities","part_time_job"]
```

```python
# PREPROCESSING TRANSFORMER
```

```python
preprocessor = ColumnTransformer(
transformers=[('num', StandardScaler(), numerical_features),('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features )]
    )
```

```python
X_train_processed = preprocessor.fit_transform(X_train)

X_test_processed = preprocessor.transform(X_test)
```

```python
# CHECK SHAPE
```

```python
print("Processed X_train shape:", X_train_processed.shape)
print("Processed X_test shape:", X_test_processed.shape)
```

```python
# CHECK MISSING VALUES
```

```python
print("Missing values in X_train:",np.isnan(X_train_processed).sum())

print("Missing values in X_test:", np.isnan(X_test_processed).sum())
```

```python
# FINAL CHECK
```

```python
print("Training data:", X_train_processed.shape)
print("Testing data:", X_test_processed.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```
